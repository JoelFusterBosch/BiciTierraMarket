## Problemes de qualitat detectats i exemples

Per a explicar els problemes de qualitat primer tenim que separar els tipus de problemes que han aparegut en els respectius fitxers csv, els quals son:

- Dades nul·les
- Dades exagerades (outliners)
- Columnes repetides

**NOTA**

Encara que en este cas els csv no tenien ni **dades exagerades (outliners)** ni **files repetides**, és important explorar les solucions als problemes esmentats en anterioritat, degut a que proporcionen informació més completa dels problemes que és pot trobar en situacions paregudes.

### Dades nul·les

Per a eliminar els valors nuls de les files tenim unes quantes solucions, les quals cadascuna d'elles és correcta segons els tipus de dades i context:

- Substituïr per `0`
- Substituïr per la mitjana (`median`)
- Eliminar la fila

Com bé s'ha esmentat abans cadascuna de les formes és útil, però en alguns casos no va a ser correcte una forma o altra.

Per exemple si n'hi ha un paràmetre `edat` la millor opció seria usar la `mitjana`, perque encara que no sabem la edat i tinguerem que triar un número, seria millor usar la `mitjana` a que tinga una edat de `0`, mentre que si n'hi ha un paràmetre `descompte` la millor opció seria cambiar el nul per `0` degut a que simplement en eixe moment no n'hi havien descomptes i al inserir les dades les havien deixades en blanc degut a la raó esmentada.

### Dades exagerades (`outliners`)

En el cas de les dades exagerades o millor conegudes com a `outliners` son dades que no són normals respecte a altres dades, el millor escenari posible per a solucionar aquest problema seria eliminar eixes files que sobreisquen del normal, per què?

Perque si deixem eixes dades que estan fora de si, els models podrien agafar eixes dades i fer calculs/prediccions incorrectes.

### Dades repetides

Aquest cas és molt simple de resoldre, en el cas que n'hi hagen dades repetides en alguna fila del csv, simplement elimina eixa fila, degut a que podria causar conflictes a l'hora d'entrenar i comparar dades en el model.

Però n'hi ha que tindre en compte, en alguns casos encara que siga a simple vista poden pareixer repetits, en ralitat tenen varis camps repetits menys 1, com podria ser el següent exemple:

| Nom   | Cognoms     | Edat |
|-------|-------------|------|
| Pep   | Grau Garcia | 30   |
| Pepet | Grau Garcia | 30   |

De totes maneres sols aplicaria si mirares les dades de forma visual, perque a l'hora de buscar repetits en `Python` usant `Pandas`, ell mateixa troba ràpida i facilment totes les dades repetides.

## Baseline, model alternatiu i mètrica que proposem

En aquesta primera fase hem fet proves exploratòries: cada membre de l'equip ha provat un model diferent per veure com es comporten les dades. Encara no són definitius ni comparables del tot entre ells (cada prova usa variables i particions diferents), però ens serveixen per decidir com continuar.

**Baseline**

Com a punt de partida, el model més simple possible: predir sempre la mitjana de `sales_units`. Qualsevol model que proposem l'ha de superar.

**Models provats**

| Model | Tipus | Resultat orientatiu |
|---|---|---|
| Regressió lineal | Regressió | R² ≈ 0,90 · MAE ≈ 11 unitats |
| Regressió polinomial | Regressió | R² ≈ 0,92 · MAE ≈ 10 unitats (amb grau alt sobreajusta) |
| Regressió logística multiclasse | Classificació (baix / mitjà / alt) | Accuracy ≈ 0,80 |
| Gradient Boosting | Regressió | R² ≈ 0,79 · MAE ≈ 16 unitats |

**Per què passem de classificació a regressió**

Amb la regressió logística hem convertit les vendes en tres classes (baixes, mitjanes, altes). Detecta bé les vendes altes (89 % de precisió), però falla més en les mitjanes (67 %): tallar un número en trams fa que els casos de la frontera siguen ambigus i perdem informació. Per això, en aquest cas, no ens convé un model categòric. Hem provat un model de regressió que prediu directament les unitats, el **Gradient Boosting**, que vam veure a classe i que va donar bons resultats.

**Model alternatiu**

Proposem treballar amb models de **regressió**, ja que prediuen el número exacte d'unitats. La regressió lineal és la base perquè és fàcil d'interpretar: ens diu quant pesa cada factor (màrqueting, descomptes, etc.). Com a alternatives, tenim la polinomial i el Gradient Boosting, que poden captar relacions no lineals.

**Mètrica**

- **MAE**: s'entén directament en unitats ("ens equivoquem unes X unitats de mitjana").
- **R²**: per comparar models entre ells.

Per a la classificació, usem l'accuracy i la matriu de confusió.

**Evidències**

## Hipòtesi 1 i hipòtesi 2

volem saber l’efecte de avg_discount i marketing_spend una vegada descomptat el que ja expliquen el mes, el canal, la línia de producte i la regió. Les nostres pregunes principals i secundàries ens duen a plantejar-nos estes hipòtesis:

0. Les vendes depenen sobretot del canal, la línia de producte, la regió i el mes de l’any.
-   D’on surt: a `prova_visualitzacions`, la mitjana per canal × línia × regió arriba a 0,85 de correlació amb les vendes, i el patró mensual es repeteix cada any.

1. A igualtat de canal, línia, regió i mes, més gasto en màrqueting augmenta les vendes, però amb rendiments decreixents.

-   Variables: marketing_spend.
-   Model: polinomial (corba còncava i pendent que baixa).
-   Es refuta si la corba és una recta o és plana.

2. A igualtat de canal, línia, regió i mes, un descompte més alt augmenta les vendes, però amb un efecte limitat.

-   Variables: avg_discount, planned_discount.
-   Model: lineal (signe i pes) i polinomial (si hi ha un punt on deixa de compensar).
-   Es refuta si el coeficient és negatiu o no significatiu.

3. L’efecte del màrqueting passa en gran part per les visites web. Més màrqueting genera més website_sessions, i les sessions generen vendes, sobretot a l’e-commerce.

-   Model: lineal, comparant el coeficient de marketing_spend amb i sense website_sessions.
-   Es refuta si el coeficient no canvia en afegir les sessions.

Les tres es poden justificar amb la matriu de correlació, però tingues en compte que és un suport feble: una correlació baixa només diu que la relació lineal és xicoteta. No diu res de si la corba és còncava (H1) ni de si el màrqueting passa per les sessions (H3). Això ho han de decidir els models, i és el que la hipòtesi ja planteja.
