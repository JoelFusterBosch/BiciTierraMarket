Per a tindre un notebook amb preparació reproduïble és té que tindre en compte unes quantes pautes a recalcar, aquests és divideixen en:

- Tindre una ordre seqüencial estricta
- No modificar els fitxers de dades (`Excel`/`csv`) originals i justificar els canvis realitzats de la versió final
- Donar un valor a la semilla 

### Ordre seqüencial estricta

L'ordre de seqüències en un notebook és molt important a l'hora de fer-se i executar-se, perque si es fa de forma incorrecta aleshores el notebook podria donar errors per botar-se certs pasos o no estar organitzat de forma correcta.

És com si tractares de llegir un llibre en un ordre diferent (Per exemple: Desenvolupament -> Inici -> Desenllaç), de llegir és pot, però no t'enteraries ni de que tracta ni de com evoluciona la història, això és un exemple que pot aplicar-se si el notebook no té un ordre esencial.

### Canvis justificats i el perque no modificar els fitxers originals

Els canvis justificats son les modificacions que es fan en els csv per a que no n'hi hagen problemes a l'hora d'entrenar models, eixes dades podrien ser:

- Dades incorrectes
- Dades exagerades
- Dades nul·les
- Dades repetides

Eixes dades és corregeixen/eliminen i és guarden en altres fitxers csv (per exemple `sales_history_revisió.csv`).

També és important **NO** modificar els fitxers originals, perque sempre té que n'hi ha que tindre els fitxers originals si la revisió ix mal o et demanen informació que en l'original estava, i en la revisió no.

### Valor de la semilla (`seed`)

El valor de la semilla l'usem per a que quan algu execute el notebook els resultats siguen sempre els mateixos, és a dir dona igual si executes la linia 1, 20 o 500 vegades, sempre eixira els mateixos resultats.

Si no s'usa aleshores cada vegada degut a que la semilla (`sucessió de números pseudoaleatoris generats pel ordenador`) va canviant cada vegada que s'executa, aleshores el valor sempre sera diferent cada vegada que s'executa.

I anem a justificar especialment l’exclusió d’ingressos i sessions del mateix mes.

| Variable | Tipus | Es coneix abans del mes? | Decisió | Justificació |
|---|---|---|---|---|
| `month` | Temporal | Sí | **Admesa** | És el calendari; permet extraure mes de l'any i estacionalitat. |
| `channel` | Categòrica | Sí | **Admesa** | Defineix la sèrie (canal). |

| `list_price` | Numèrica | Sí | **Admesa** | Preu de llista, conegut abans del mes. |

| `sales_units` (mes *t*) | Numèrica | No | **Exclosa com a predictor (és l'objectiu)** | És la variable a predir. |

| `marketing_spend` (mes *t*) | Numèrica | No, només en tancar el mes | **Exclosa** | És la despesa real; abans del mes només es coneix el pressupost (`marketing_budget`). |
| `avg_discount` (mes *t*) | Numèrica | No, només en tancar el mes | **Exclosa** | És el descompte real; abans del mes només es coneix `planned_discount`. A més, està ocult en algunes files. |
| `closing_stock_units` (mes *t*) | Numèrica | No, només en tancar el mes | **Exclosa** | Es calcula amb les vendes: `stock_units − sales_units` (sense reposició). Revela la resposta. |
| `stockout` (mes *t*) | Binària | No, només en tancar el mes | **Exclosa** | Depén de si les vendes han esgotat l'estoc; és conseqüència de l'objectiu. |

#### Justificació de l'exclusió de `revenue` i `website_sessions`

##### `revenue` (ingressos del mes)
- Fuga directa de l’objectiu. Per definició, `revenue = sales_units × list_price × (1 − avg_discount)`. Com que `list_price` és conegut i `avg_discount` es pot aproximar amb `planned_discount`, es pot aïllar `sales_units ≈ revenue / (list_price × (1 − descompte))`. El model no “predirien”: **aclariria l’objectiu**.

- **No està disponible en el moment de la decisió**. Els ingressos es consoliden en tancar el mes. En producció, quan haguem de predir (abans de començar el mes), eixa dada encara no existeix.

- **Resultat enganyós**. Es veurien mètriques d’error excel·lents en validació que no es podrien reproduir en la realitat.

- **Cas permès**. Només l’revenue de mesos anteriors (retards) és vàlid.

#### `website_sessions` (sessions web del mes)

- **Simultaneïtat temporal**. Les sessions s’acumulen durant el mes i són, en part, conseqüència de les mateixes forces que generen les vendes (campanya, descomptes, demanda). Només es coneixen en tancar el mes.

- **Causalitat inversa**. Més vendes i més trànsit van junts; usar les sessions del mes seria utilitzar un “efecte” contemporani de la demanda per predir la demanda.

- **No és planificable**. A diferència de marketing_budget o planned_discount, que decideix l’empresa, les sessions no es poden fixar ni conéixer per endavant, de manera que no es poden aportar al model en el moment de predir.

- **Cas permès**. Les sessions dels mesos anteriors (p. ex. t−1) sí que són un senyal vàlid de l’interés recent.

## <span style="color:#16A085">Resultats de les referències</span>

Una **referència** (*baseline*) no és un model entrenat: és una **regla simple**. Qualsevol model que entrenem a la setmana 3 l'ha de millorar. Si no, no aporta res.

- **Mitjana històrica de grup:** "aquest mes vendrem el que ven de mitjana aquesta sèrie". La mitjana de cada sèrie (canal + producte + regió) es calcula **només amb entrenament**.
- **Mateix mes de l'any anterior:** "aquest mes vendrem el mateix que fa un any". És directament `lag_12`.

Seguim el mateix procediment que el notebook [`02_modelatge_baseline.ipynb`](../notebooks/02_modelatge_baseline.ipynb):

- Entrenament 2022–2024 (2021 queda fora perquè no té `lag_12`) i validació 2025. La prova (2026) no es toca.
- Les prediccions es limiten entre 0 i l'estoc disponible, perquè no es poden vendre més bicicletes de les que hi ha.

Mètriques en validació:

- **MAE**: error mitjà en **unitats** per sèrie i mes.
- **WAPE**: suma dels errors / suma de les vendes reals, en **%**.

```python
from pathlib import Path
import numpy as np
import pandas as pd

DATA = Path('..') / 'dades'
KEY = ['channel', 'product_line', 'region']

sales = pd.read_csv(DATA / 'sales_history.csv')
sales['month'] = pd.to_datetime(sales['month'], format='%Y-%m')
sales = sales.sort_values(KEY + ['month']).reset_index(drop=True)

# Retards dins de cada sèrie
for lag in [1, 12]:
    sales[f'lag_{lag}'] = sales.groupby(KEY)['sales_units'].shift(lag)
usable = sales.dropna(subset=['lag_1', 'lag_12'])

train = usable[usable['month'] < '2025-01-01']
valid = usable[(usable['month'] >= '2025-01-01') & (usable['month'] < '2026-01-01')]

def score(y, pred):
    y, pred = np.asarray(y, dtype=float), np.asarray(pred, dtype=float)
    return {'MAE (unitats)': np.mean(np.abs(y - pred)),
            'WAPE (%)': 100 * np.abs(y - pred).sum() / y.sum()}

def cap(frame, pred):
    return np.clip(pred, 0, frame['stock_units'].to_numpy())

# Referència 1: mitjana de cada sèrie, només amb entrenament
group_mean = train.groupby(KEY)['sales_units'].mean().rename('pred').reset_index()
pred_mean = valid[KEY].merge(group_mean, on=KEY, how='left')['pred'].to_numpy()

# Referència 2: mateix mes de l'any anterior
pred_lag12 = valid['lag_12'].to_numpy()

pd.DataFrame([
    {'referència': 'Mitjana històrica de grup', **score(valid['sales_units'], cap(valid, pred_mean))},
    {'referència': "Mateix mes de l'any anterior", **score(valid['sales_units'], cap(valid, pred_lag12))},
]).round(2)
```

| Referència | MAE (unitats) | WAPE (%) |
|---|---|---|
| Mitjana històrica de grup | 18,75 | 18,06 |
| **Mateix mes de l'any anterior** | **17,40** | **16,77** |

### <span style="color:#D68910">Què implica per al negoci</span>

- La millor referència és **el mateix mes de l'any anterior**. Guanya perquè recull l'**estacionalitat** (estiu, Nadal…), mentre que la mitjana dona el mateix valor tot l'any.
- Una sèrie ven de mitjana unes **104 unitats al mes**. Un MAE de 17 vol dir que, sense cap model, **cada mes sobren o falten unes 17 bicicletes per sèrie** (≈ 17 % del volum).
- Si es demana de més: estoc parat i diners immobilitzats. Si es demana de menys: **ruptures d'estoc** i vendes perdudes.
- **Llistó per a la setmana 3:** el model que proposem ha de baixar d'un **MAE de 17,40 unitats / WAPE del 16,77 %** amb les mateixes files de validació.

# <span style="color:#2E86C1">Un exemple de com una mala imputació o una variable posterior podria falsejar el resultat.</span>

Un exemple de mala imputació seria en el dataset de vendes (`sales_history.csv`) podria ser:

Que en la columna `sales_units` tinga un valor nul i en eixe cas enlloc de ficar la mitjana (`mean`) ficarem `0` aleshores estariem deduïnt que en eixe mes no s'han fet vendes, cosa que seria completament erronia.

Perque veient les propies dades del csv veiem que en ningun cas les vendes son tan baixes que arriven a `0`, aleshores enlloc de solucionar el problema aleshores allí estariem creant una dada exagerada (`outliner`), la qual a l'hora de fer prediccions podria alterar les dades de tal forma que les prediccions s'allunyen molt de les dades que realment voldriem arribar.

Amb la conseqüència de que el `MAE` (**M**ean **A**bsolute **E**rror) i el `MSE` (**M**ean **S**quared **E**rror) augmenten i el `R2` (`Coeficient de determinació`) disminuïsca.
