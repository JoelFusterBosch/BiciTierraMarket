# Un exemple de com una mala imputació o una variable posterior podria falsejar el resultat. 

Un exemple de mala imputació seria en el dataset de vendes (`sales_history.csv`) podria ser:

Que en la columna `sales_units` tinga un valor nul i en eixe cas enlloc de ficar la mitjana (`mean`) ficarem `0` aleshores estariem deduïnt que en eixe mes no s'han fet vendes, cosa que seria completament erronia.

Perque veient les propies dades del csv veiem que en ningun cas les vendes son tan baixes que arriven a `0`, aleshores enlloc de solucionar el problema aleshores allí estariem creant una dada exagerada (`outliner`), la qual a l'hora de fer prediccions podria alterar les dades de tal forma que les prediccions s'allunyen molt de les dades que realment voldriem arribar.

Amb la conseqüència de que el `MAE` (**M**ean **A**bsolute **E**rror) i el `MSE` (**M**ean **S**quared **E**rror) augmenten i el `R2` (`Coeficient de determinació`) disminuïsca.
